```python
raw_data = np.array([
    [101.0,  28.0,  50000.0,  1.0],
    [102.0,  35.0,  62000.0,  2.0],
    [103.0, np.nan,  45000.0,  1.0],  # Missing Age
    [104.0,  42.0, 999999.0,  3.0],  # Outlier Salary
    [102.0,  35.0,  62000.0,  2.0],  # Duplicate row
    [105.0, -25.0,  55000.0,  1.0],  # Invalid negative Age
    [106.0,  50.0,  80000.0,  2.0]
])

np.set_printoptions(suppress=True, precision=1)
print("Raw Data:\n", raw_data)
```

```python
clean_data=np.unique(raw_data,axis=0)
```

```python
age_col=clean_data[:, 1]
age_nan=np.isnan(age_col)

mean_age=np.nanmean(age_col)

age_col[age_nan]=np.round(mean_age,1)
clean_data[:,1]=age_col
```

```python
clean_data[:,1]=np.abs(clean_data[:,1])
```

```python
clean_data
```

```python
sal_col=clean_data[:,2]
med_sal=np.median(sal_col[sal_col<150000])

sal_col[sal_col>150000]=med_sal
clean_data[:,2]=sal_col
```

```python
emp_id=clean_data[:,0].astype(int)
emp_id
```

### Problem 1: Top Revenue Drivers by Location Tier
Question: What are the top 3 best-selling product categories within each Outlet Location Type (Tier 1, Tier 2, Tier 3)?

```python
original_df.head(5)
```

```python
type_sales=original_df.groupby(['Outlet Location Type','Item Type'])['Sales'].sum().reset_index()
top_3_categories=type_sales.sort_values(['Outlet Location Type','Sales'], ascending=[True,False]).groupby('Outlet Location Type').head(3)
print(top_3_categories)
```

```python
original_df.info()
```

### Problem 2: Impact of Item Visibility on Sales
Question: Does higher product placement/visibility (Item Visibility) result in higher sales?

```python
original_df['Item Visibility Bin']=pd.qcut(original_df['Item Visibility'],q=5,labels=['Very Low','Low','Medium','High','Very High'])
impact=original_df.groupby('Item Visibility Bin')['Sales'].agg(['mean','median','count']).reset_index()
print(impact)
```

### Problem 3: Store Maturity vs. Performance
Question: Does the age of an outlet (Outlet Establishment Year) significantly impact its average sales?

```python
original_df['Outlet Age']=2026-original_df['Outlet Establishment Year']
age_indicators=original_df.groupby(['Outlet Establishment Year','Outlet Age'])['Sales'].agg(['mean','count']).reset_index()
print(age_indicators)
```

### Problem 4: Customer Preference (Low Fat vs Regular) Across Outlet Types
Question: Do customers purchasing at Grocery Store vs. Supermarket Type3 prefer Low Fat or Regular items?

```python
fat_preferance=original_df.groupby(['Item Fat Content','Outlet Type'])['Sales'].agg(['mean','count']).unstack()
print(fat_preferance)
```

### Problem 5: Identifying Underperforming Outlets
Question: Which specific outlet locations (Outlet Identifier) are underperforming in total revenue and order volume?

```python
underperformer=original_df.groupby('Outlet Identifier').agg(Total_Sales=('Sales','sum'),Avg_Rating=('Rating','mean'),Item_Count=('Outlet Identifier','count')).reset_index().sort_values('Total_Sales',ascending=False)
print(underperformer)
```
